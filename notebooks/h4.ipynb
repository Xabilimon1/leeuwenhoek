```python
# 2. Cargar la red entrenada y los números del gemelo diseñado
import ipywidgets as W
import matplotlib.pyplot as plt

from leeuwenhoek.design import designed_weights, read_off
from leeuwenhoek.exact import exact_accuracy
from leeuwenhoek.models import load_argmedian
from leeuwenhoek.plots import overview

_, trained = load_argmedian(4, 3)
p = read_off(trained)
print(f"red entrenada: acierta el {100 * exact_accuracy(trained)[0]:.2f} % (exacto)")
```

## Mira dentro de la red

```python
# 3. Controles
def show(red, x1, x2, x3, memoria):
    w = trained if red == "entrenada" else designed_weights(p["a"], memoria, p["e"], p["B2"], p["B3"])
    print(f"red {red}: acierta el {100 * exact_accuracy(w)[0]:.2f} % de las veces (exacto)")
    overview(w, [x1, x2, x3])
    plt.show()


slider = dict(min=-2, max=2, step=0.005, continuous_update=False, readout_format=".3f")
W.interact(
    show,
    red=W.ToggleButtons(options=["entrenada", "diseñada"]),
    x1=W.FloatSlider(value=0.9, description="x₁", **slider),
    x2=W.FloatSlider(value=-0.4, description="x₂", **slider),
    x3=W.FloatSlider(value=0.5, description="x₃", **slider),
    memoria=W.FloatSlider(value=round(p["b"], 2), min=0.02, max=1.5, step=0.01, continuous_update=False,
                          description="memoria"),
);
```
